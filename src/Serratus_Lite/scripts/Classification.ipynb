```python
import pandas as pd
import matplotlib
import numpy as np
%matplotlib inline

pd.set_option('display.max_columns', None)
```

```python
ZfPV_df = pd.read_csv('../results/ZfPV/ZfPV.parsed.csv', index_col='SRA_run_accession')
ZfPV_df
# ZfPV_df['cov x depth'] = ZfPV_df['famcvg'] * ZfPV_df['depth']
```

```python
zpico_df = pd.read_csv('../results/zpico/zpico.parsed.csv', index_col='SRA_run_accession')
zpico_df
```

```python
zpoxv_df = pd.read_csv('../results/zpoxv/zpoxv.parsed.csv', engine='python', index_col='SRA_run_accession')
zpoxv_df
```

```python
aln_df = pd.concat([ZfPV_df['aln'], zpico_df['aln'], zpoxv_df['aln']],
                      axis=1,
                      keys=['ZfPV', 'zpico', 'zpoxv'])
# aln_df = aln_df.fillna(0)
aln_df
```

```python
zf_counts_df = pd.read_csv('../zf_counts.csv', index_col='GeneID')
zf_counts_df.index.name = 'SRA_run_accession'
zf_counts_df
```

```python
values_df = aln_df.merge(zf_counts_df, left_on='SRA_run_accession', right_on='SRA_run_accession', how='inner')
values_df
```

```python
normed_values_df = values_df.drop('reads', axis=1).multiply(10e+6).div(values_df['reads'], axis=0)
normed_values_df
```

```python
class_df = normed_values_df.copy()
labels = ['Low', 'Medium', 'High', 'Very High']
bin_dict = {}

# class_df['ZfPV'] = pd.cut(class_df['ZfPV'],
#                        bins=[-1, 0, 1.5, 10, np.inf],
#                        labels=labels,
#                        include_lowest=False,
#                        right=True)

# class_df['zpico'] = pd.cut(class_df['zpico'],
#                        bins=[-1, 0, 3, 20, np.inf],
#                        labels=labels,
#                        include_lowest=False,
#                        right=True)

# class_df['zpoxv'] = pd.cut(class_df['zpoxv'],
#                        bins=[-1, 0, 2, 30, np.inf],
#                        labels=labels,
#                        include_lowest=False,
#                        right=True)
for col in class_df.columns:
    class_df[col], bin_dict[col] = pd.qcut(class_df[col], q=4, labels=labels, retbins=True)
    class_df[col] = class_df[col].cat.add_categories('UND').fillna('UND')
for key, val in bin_dict.items():
    print(f'{key} bins: {val}')
class_df
```

```python
ZfPV_nonzero = pd.DataFrame(normed_values_df[normed_values_df['ZfPV'] != 0]['ZfPV'])
plt = ZfPV_nonzero.plot.hist(bins=range(50))
plt.vlines(bin_dict['ZfPV'][: -1], ymin=0, ymax=300, color='orange', linestyle='dashed')
plt
ZfPV_nonzero
# bins:
```

```python
zpico_nonzero = pd.DataFrame(normed_values_df[normed_values_df['zpico'] != 0]['zpico'])
plt = zpico_nonzero.plot.hist(logx=False, bins=range(50))
plt.vlines(bin_dict['zpico'][: -1], ymin=0, ymax=200, color='orange', linestyles='dashed')
plt
zpico_nonzero
# bins:
```

```python
zpoxv_nonzero = pd.DataFrame(normed_values_df[normed_values_df['zpoxv'] != 0]['zpoxv'])
plt = zpoxv_nonzero.plot.hist(bins=(range(50)))
plt.vlines(bin_dict['zpoxv'][: -1], ymin=0, ymax=200, color='orange', linestyles='dashed')
plt
# bins:
```

```python
bool_df = normed_values_df.astype(bool)
bool_df['Any'] = bool_df.any(axis=1)
bool_df
```
