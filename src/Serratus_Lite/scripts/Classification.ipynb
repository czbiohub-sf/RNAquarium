```python
import pandas as pd
import matplotlib
import numpy as np
%matplotlib inline
```

```python
ZfPV_df = pd.read_csv('../results/ZfPV/ZfPV.parsed.csv')
ZfPV_df
```

```python
zpico_df = pd.read_csv('../results/zpico/zpico.parsed.csv')
zpico_df
```

```python
zpoxv_df = pd.read_csv('../results/zpoxv/zpoxv.parsed.csv', engine='python')
zpoxv_df
```

```python
values_df = pd.concat([ZfPV_df['readlength'], zpico_df['readlength'], zpoxv_df['readlength']],
                      axis=1,
                      keys=['ZfPV', 'zpico', 'zpoxv'])
values_df
```

```python
ZfPV_nonzero = pd.DataFrame(values_df[values_df['ZfPV'] != 0]['ZfPV'])
ZfPV_nonzero.plot.hist()
ZfPV_nonzero
# bins:
# x = 0
# 0 < x <= 60
# 60 < x <= 90
# 90 < x <= 120
# x > 120
```

```python
zpico_nonzero = pd.DataFrame(values_df[values_df['zpico'] != 0]['zpico'])
zpico_nonzero.plot.hist(xticks=range(0, 350, 25))
zpico_nonzero
# bins:
# x = 0
# 0 < x <= 60
# 60 < x <= 90
# 90 < x <= 120
# x > 120
```

```python
zpoxv_nonzero = pd.DataFrame(values_df[values_df['zpoxv'] != 0]['zpoxv'])
zpoxv_nonzero.plot.hist()
zpoxv_nonzero
# bins:
# x = 0
# 0 < x <= 60
# 60 < x <= 90
# 90 < x <= 120
# x > 120
```

```python
class_df = values_df.copy()
bins = [0, 1, 61, 91, 121, np.inf]
labels = ['Nothing', 'Low', 'Medium', 'High', 'Very High']
for col in class_df.columns:
    class_df[col] = pd.cut(class_df[col],
                           bins=bins,
                           labels=labels,
                           include_lowest=True,
                           right=False)
class_df
```

```python
bool_df = values_df.astype(bool)
bool_df['Any'] = bool_df.any(axis=1)
bool_df
```

```python
values_df.to_csv('../matrices/values.csv', index=False)
class_df.to_csv('../matrices/multiclass.csv', index=False)
bool_df.to_csv('../matrices/boolean.csv', index=False)
```
