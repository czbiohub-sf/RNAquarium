```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib notebook
```

```python
df = pd.read_csv('parsed_reads.csv')
```

```python
df
```

```python
df = df[df['Mate2'].notna()]
```

```python
fig, ax = plt.subplots()
ax.set(xlim=(0, 175), ylim=(0, 175))
sns.kdeplot(df[['Mate1','Mate2']],
            x='Mate1',
            y='Mate2',
            fill=True,
            ax=ax)
plt.show()
```
